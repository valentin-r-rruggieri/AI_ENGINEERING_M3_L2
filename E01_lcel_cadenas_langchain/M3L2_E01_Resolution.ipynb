# M3L2 E01 - LCEL y cadenas en LangChain

## Objetivo

Vas a unir componentes con LCEL:

```python
chain = prompt | llm | parser
```

## Tabla conceptual

| Componente | Entrada | Salida |
|---|---|---|
| `ChatPromptTemplate` | diccionario con variables | mensajes para chat |
| `ChatOpenAI` | mensajes | `AIMessage` |
| `StrOutputParser` | `AIMessage` | string |

## Diagrama LCEL

```text
{"tema": "cocina", "concepto": "sushi"}
        |
        v
ChatPromptTemplate -> ChatOpenAI -> StrOutputParser -> string final
```

````python
import os
import getpass
from langchain_openai import ChatOpenAI

if not os.getenv("OPENAI_API_KEY"):
    os.environ["OPENAI_API_KEY"] = getpass.getpass("Ingresa tu OpenAI API key: ")


def obtener_modelo(temperature: float = 0.2):
    return ChatOpenAI(model="gpt-4o-mini", temperature=temperature)


print("API key cargada en la variable de entorno OPENAI_API_KEY")
````

````python
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.output_parsers import StrOutputParser

llm = obtener_modelo()
parser = StrOutputParser()
````

````python
prompt = ChatPromptTemplate.from_messages([
    ("system", "Sos un docente de AI Engineering. Respondes claro, breve y en castellano."),
    ("human", "Explica el concepto {concepto} usando un ejemplo del tema {tema}."),
])
print(prompt.input_variables)
````

````python
cadena = prompt | llm | parser
print(cadena)
````

````python
resultado = cadena.invoke({"tema": "cocina", "concepto": "sushi"})
print(resultado)
````

````python
for ejemplo in [
    {"tema": "programacion", "concepto": "funcion"},
    {"tema": "atencion al cliente", "concepto": "ticket"},
    {"tema": "marketing", "concepto": "segmentacion"},
]:
    print("\n---", ejemplo, "---")
    print(cadena.invoke(ejemplo))
````

## Errores comunes

| Error | Causa probable | Como pensarlo |
|---|---|---|
| Falta una variable | El dict no trae todos los campos | Revisar `prompt.input_variables` |
| El resultado no es string | Falta `StrOutputParser()` | El modelo devuelve `AIMessage` |
| La chain falla | Alguna pieza es `None` | Completar TODOs antes de componer |

````python
assert prompt is not None
assert cadena is not None
assert isinstance(resultado, str)
print("Checks OK")
````
