# M3L2 E04 - StrOutputParser (Resolution)


```python
import os, getpass
if not os.getenv("OPENAI_API_KEY"):
    os.environ["OPENAI_API_KEY"] = getpass.getpass("OpenAI API key: ")

from langchain_openai import ChatOpenAI
from langchain_core.output_parsers import StrOutputParser

llm = ChatOpenAI(model="gpt-4o-mini", temperature=0)
parser = StrOutputParser()
```

```python
# Sin parser: AIMessage
result_raw = llm.invoke("Di solo: hola")
print(f"Sin parser: tipo={type(result_raw).__name__}, content='{result_raw.content}'")
```

```python
# El parser aplicado directamente
result_parsed = parser.invoke(result_raw)
print(f"Con parser directo: tipo={type(result_parsed).__name__}, valor='{result_parsed}'")
```

```python
# TODO 1: componer llm | parser
chain_con_parser = llm | parser
print(f"Chain: {type(chain_con_parser).__name__}")
```

```python
# TODO 2: invocar y comparar
result_chain = chain_con_parser.invoke("Di solo: hola")
print(f"Con llm solo:     tipo={type(result_raw).__name__}")
print(f"Con llm | parser: tipo={type(result_chain).__name__}")
print(f"Valor: '{result_chain}'")
```

```python
def run_checks():
    from langchain_core.messages import AIMessage
    assert isinstance(llm.invoke("test"), AIMessage)
    result = chain_con_parser.invoke("test")
    assert isinstance(result, str) and len(result) > 0
    print("M3L2 E04 Resolution checks passed")

run_checks()
```
