# M3L2 E04 - RAG basico

## Objetivo

RAG significa Retrieval Augmented Generation. Primero recuperamos contexto relevante y despues generamos la respuesta.

## Flujo RAG

```text
Documento -> Splitter -> Chunks -> Embeddings -> Chroma -> Retriever -> Contexto -> Prompt -> Modelo
```

## Componentes

| Componente | Responsabilidad |
|---|---|
| Loader | Cargar texto |
| Splitter | Dividir en fragmentos |
| Embeddings | Convertir texto a vectores |
| Chroma | Guardar vectores |
| Retriever | Recuperar fragmentos relevantes |
